```python
import os
import shutil
import pandas as pd
from pathlib import Path

# Configure these paths
BASE_PATH = "../split_node21_sets/crop"
# BASE_PATH = "../split_node21_sets/lung_seg"
# BASE_PATH = "../split_node21_sets/arch_seg"  # Path to your arch_seg folder
MAPPING_CSV = "../split_node21_sets/mapping.csv"  # Path to your mapping CSV file

print("Directory Reorganization Script")
print("=" * 40)

# Read the mapping CSV
try:
    mapping_df = pd.read_csv(MAPPING_CSV)
    print(f"Loaded mapping CSV with {len(mapping_df)} entries")
except FileNotFoundError:
    print(f"Error: Could not find mapping CSV at {MAPPING_CSV}")
    exit()
except Exception as e:
    print(f"Error reading CSV: {e}")
    exit()

# Verify required columns exist
if 'dir_name' not in mapping_df.columns or 'orig_dataset' not in mapping_df.columns:
    print("Error: CSV must contain 'dir_name' and 'orig_dataset' columns")
    exit()

# Create a dictionary for quick lookups
dir_to_dataset = dict(zip(mapping_df['dir_name'], mapping_df['orig_dataset']))

base_path = Path(BASE_PATH)

# Check if base path exists
if not base_path.exists():
    print(f"Error: Base path {base_path} does not exist")
    exit()

# Show current structure
print("\nCurrent directory structure:")
for split in ['train', 'test']:
    split_path = base_path / split
    if split_path.exists():
        for category in ['nodule', 'normal']:
            category_path = split_path / category
            if category_path.exists():
                dirs = [d.name for d in category_path.iterdir() if d.is_dir()]
                print(f"  {category_path}: {len(dirs)} directories")
                if dirs:
                    print(f"    Examples: {dirs[:3]}{'...' if len(dirs) > 3 else ''}")

# Process both train and test folders
for split in ['train', 'test']:
    split_path = base_path / split
    if not split_path.exists():
        print(f"Warning: {split_path} does not exist, skipping...")
        continue
        
    # Process both nodule and normal folders
    for category in ['nodule', 'normal']:
        category_path = split_path / category
        if not category_path.exists():
            print(f"Warning: {category_path} does not exist, skipping...")
            continue
        
        print(f"\nProcessing {category_path}...")
        
        # Get all directories in this category folder
        for item in category_path.iterdir():
            if item.is_dir():
                dir_name = item.name
                
                # Check if this directory is in our mapping
                if dir_name in dir_to_dataset:
                    orig_dataset = dir_to_dataset[dir_name]
                    
                    # Create the new path structure
                    new_base = base_path / orig_dataset / split / category
                    new_path = new_base / dir_name
                    
                    # Create the new directory structure if it doesn't exist
                    new_base.mkdir(parents=True, exist_ok=True)
                    
                    try:
                        # Move the entire directory
                        shutil.move(str(item), str(new_path))
                        print(f"Moved: {item} -> {new_path}")
                    except Exception as e:
                        print(f"Error moving {item}: {e}")
                else:
                    print(f"Warning: Directory '{dir_name}' not found in mapping CSV")

print("\nReorganization complete!")

# Show new structure
print("\nNew directory structure:")
for item in base_path.iterdir():
    if item.is_dir() and item.name not in ['train', 'test']:
        print(f"Dataset: {item.name}")
        for split in ['train', 'test']:
            split_path = item / split
            if split_path.exists():
                for category in ['nodule', 'normal']:
                    category_path = split_path / category
                    if category_path.exists():
                        count = len([d for d in category_path.iterdir() if d.is_dir()])
                        print(f"  {split}/{category}: {count} directories")
```
